### 2) distplot : histogram + density plot

https://seaborn.pydata.org/generated/seaborn.distplot.html

* hist_kws = dict() : 히스토그램 꾸미기 옵션, 딕셔너리 형태로 입력.
* 히스토그램과 밀도함수 그래프를 겹쳐서 표현



① titanic['Fare'] 에 대해서 distplot 그리기
* bin의 개수를 10 ~ 50 사이에서 조절
* 히스토그램 꾸미기. hist_kws = dict(edgecolor='grey')) 또는 hist_kws = {'edgecolor':'grey'}

```python
sns.distplot(titanic['Age'], bins =20, hist_kws = {'edgecolor':'grey'}) # 딕셔너리 형태로 옵션 전달
plt.show()

# hist_kws : Seaborn의 sns.distplot() 또는 sns.histplot() 함수에서 히스토그램의 모양과 스타일을 세밀하게 조정할 수 있는 옵션
# matplotlib의 hist() 함수에서 사용할 수 있는 옵션 활용 가능
# edgecolor: 히스토그램 막대의 테두리 색상 설정
# alpha: 막대의 투명도 설정. (0에서 1 사이의 값, 0은 완전히 투명, 1은 불투명)
# linewidth: 막대 테두리 두께 설정
# color: 막대 내부의 색상 설정
```

② boston['medv'] 집값 - distplot
* bin의 개수 10 ~ 50 사이 조절
* 히스토그램을 꾸미기. hist_kws = {'edgecolor':'black'}

```python
sns.distplot(boston['medv'], bins = 20, hist_kws = {'edgecolor':'black'})
plt.show()
```

### 7) heatmap : 두 범주 집계 시각화

https://seaborn.pydata.org/generated/seaborn.heatmap.html

* 두 범주를 집계한 결과를 색의 농도로 표현
* 집계(groupby)와 피봇(pivot) 필요
* 여러 범주를 갖는 변수 비교 시 유용


① titanic 데이터 - Embarked별, Pclass별 평균 탑승객수 집계 .count()

```python
temp1 = titanic.groupby(['Embarked','Pclass'], as_index = False)['PassengerId'].count()
temp1
```

② 'Embarked', 'Pclass', 'PassengerId'로 피봇

```python
temp2 = temp1.pivot(index='Embarked', columns='Pclass', values='PassengerId')
temp2
```

③ heatmap


```python
sns.heatmap(temp2, annot = True)
# sns.heatmap(temp2, annot = False)
plt.show()
```

```python
#  그림 크기 조정, 구간 간격 확장
plt.figure(figsize = (12,6))
sns.heatmap(temp2, annot = True, fmt = 'd', linewidth = .2)
plt.show()
```
