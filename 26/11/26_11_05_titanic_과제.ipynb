```python
def solution(slice, n):
    answer = n // slice
    if slice < n:
        return answer + 1
    else:
        return answer

solution(4, 12)
```

```python
import pandas as pd
import seaborn as sns

df = sns.load_dataset('titanic')
```

```python
# 1. 데이터 프레임 요약 정보
print("="*5, "1. 데이터 프레임 요약 정보", "=" * 5)
print("\n[각 열의 데이터 타입]")
print(df.dtypes)
print()
print("[데이터 형식 및 결측치 확인]")
df.info()
print()
print("\n[데이터 분석용 요약 통계]")
print(df.describe(include='all'))
```

```python
# 2. 데이터 정렬 및 필터링
# Fare(요금) 기준 내림차순 정렬 후 상위 10명 출력
print("=== 2. 데이터 정렬 및 필터링 (Fare 기준 내림차순 상위 10명) ===")
sorted_df = df.sort_values(by='fare', ascending=False)
print(sorted_df.loc[:, ['survived', 'pclass', 'sex', 'age', 'fare']].head(10))
```

```python
# 3. 열 추가 및 변환
# sibsp: 형제자매/배우자 수, parch: 부모/자녀 수
# 1인당 요금 = 총 요금 / (동반자 수 + 본인(1))
df['Fare_per_person'] = df['fare'] / (df['sibsp'] + df['parch'] + 1)
print("=== 3. 열 추가 및 변환 (Fare_per_person 추가 후 상위 5개 행) ===")
print(df[['fare', 'sibsp', 'parch', 'Fare_per_person']].head())
```

```python

# 4. 승객 생존율 - Pclass
# pclass에 따른 survived(생존율) 계산
print("=== 4. Pclass(티켓 클래스)에 따른 생존율 ===")
survival_rate_by_pclass = df.groupby('pclass')['survived'].mean()
print(survival_rate_by_pclass)
```
